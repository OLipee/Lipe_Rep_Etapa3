```python
df = pd.read_excel("Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx", engine="openpyxl")
df
```

```python
tabela_tratada = df.iloc[1:35, : ]
tabela_tratada
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
tabela_regiao = tabela_tratada[tabela_tratada.iloc[:, 0].isin(regioes)].copy()
```
